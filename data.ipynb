# Numerical columns

```python
numerical_columns=df.select_dtypes(include=np.number).columns
```

```python
df[numerical_columns]=df[numerical_columns].fillna(df[numerical_columns].median())
```

```python
df[numerical_columns]
```

# Categorial columns

```python
categorial_columns=df.select_dtypes(include="object").columns
```

```python
categorial_columns
```

```python
for col in categorial_columns:
    if df[col].isnull().any():
        df[col]=df[col].fillna(df[col].mode()[0])
```

# Remove duplicate records

```python
print("Duplicate:",df.duplicated().sum())
```

```python
df=df.drop_duplicates()
```

```python
print("duplicates after cleaning:",df.duplicated().sum())
```

# clean column name

```python
df.columns=(
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ","_")
    .str.replace(r"[^a-z0-9_]","",regex=True)
)

print(df.columns)
```

# clean categorial data

```python
categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:df[col] = (df[col].str.strip().str.replace(r"\s+", " ", regex=True))
```

# Validate datatypes

```python
print(df.dtypes)
```

```python
numeric_columns = [
    "prevalence_rate_",
    "incidence_rate_",
    "mortality_rate_",
    "population_affected",
    "healthcare_access_",
    "doctors_per_1000",
    "hospital_beds_per_1000",
    "average_treatment_cost_usd",
    "recovery_rate_",
    "dalys",
    "improvement_in_5_years_",
    "per_capita_income_usd",
    "education_index",
    "urbanization_rate_"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df
```

# Identity invalid values

```python
invalid_values = {
    "year": ~df["year"].between(2000, 2024),
    "prevalence_rate_": ~df["prevalence_rate_"].between(0, 100),
    "incidence_rate_": ~df["incidence_rate_"].between(0, 100),
    "mortality_rate_": ~df["mortality_rate_"].between(0, 100),
    "healthcare_access_": ~df["healthcare_access_"].between(0, 100),
    "recovery_rate_": ~df["recovery_rate_"].between(0, 100),
    "education_index": ~df["education_index"].between(0, 1),
    "urbanization_rate_": ~df["urbanization_rate_"].between(0, 100),
    "population_affected": df["population_affected"] < 0,
    "average_treatment_cost_usd": df["average_treatment_cost_usd"] < 0,
    "dalys": df["dalys"] < 0
}

for column, condition in invalid_values.items():
    print(column, ":", condition.sum(), "invalid records")
```

# Detect outliers using IQR

```python
outlier_columns = [
    "prevalence_rate_",
    "incidence_rate_",
    "mortality_rate_",
    "average_treatment_cost_usd",
    "per_capita_income_usd",
    "population_affected"
]

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(col, "Outliers:", len(outliers))
```

# Visualize the outliers

```python
plt.figure(figsize=(10,5))
sns.boxplot(x=df["average_treatment_cost_usd"])
plt.title("Treatment cost - outlier detection")
plt.show()
```
